```python
import os
from glob import glob
import pandas as pd
import cv2
import numpy as np
from onekey_algo import get_param_in_cwd

root = get_param_in_cwd('radio_dir')
label_data = pd.read_csv(os.path.join(root, 'label.csv'))
# label_data['group'] = 'train'
label_data
```

```python
samples = pd.DataFrame(glob(os.path.join(root, 'images', '*')), columns=['fpath'])
samples['ID'] = samples['fpath'].map(lambda x: os.path.basename(x).split('.')[0] + '.nii.gz')
samples['fname'] = samples['fpath'].map(lambda x: os.path.basename(x))
samples
```
